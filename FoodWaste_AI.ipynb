```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Features
features = [
    "Country",
    "Year",
    "Food Category",
    "Total Waste (Tons)",
    "Avg Waste per Capita (Kg)",
    "Population (Million)",
    "Household Waste (%)"
]

target = "Economic Loss (Million $)"

X = df[features]
y = df[target]

categorical_features = [
    "Country",
    "Food Category"
]

numerical_features = [
    "Year",
    "Total Waste (Tons)",
    "Avg Waste per Capita (Kg)",
    "Population (Million)",
    "Household Waste (%)"
]

print("Features:", features)
print("Target:", target)
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        ),
        (
            "numerical",
            StandardScaler(),
            numerical_features
        )
    ]
)

print("Preprocessor created successfully.")
```

```python
linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)

linear_model.fit(X_train, y_train)

linear_predictions = linear_model.predict(X_test)

linear_mae = mean_absolute_error(
    y_test,
    linear_predictions
)

linear_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        linear_predictions
    )
)

linear_r2 = r2_score(
    y_test,
    linear_predictions
)

print("Linear Regression")
print("------------------")
print("MAE:", round(linear_mae, 4))
print("RMSE:", round(linear_rmse, 4))
print("R²:", round(linear_r2, 4))
```

```python
random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

random_forest_model.fit(X_train, y_train)

rf_predictions = random_forest_model.predict(X_test)

rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_predictions
    )
)

rf_r2 = r2_score(
    y_test,
    rf_predictions
)

print("Random Forest")
print("------------------")
print("MAE:", round(rf_mae, 4))
print("RMSE:", round(rf_rmse, 4))
print("R²:", round(rf_r2, 4))
```

```python
gradient_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            GradientBoostingRegressor(
                n_estimators=200,
                learning_rate=0.05,
                max_depth=3,
                random_state=42
            )
        )
    ]
)

gradient_model.fit(X_train, y_train)

gradient_predictions = gradient_model.predict(X_test)

gradient_mae = mean_absolute_error(
    y_test,
    gradient_predictions
)

gradient_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        gradient_predictions
    )
)

gradient_r2 = r2_score(
    y_test,
    gradient_predictions
)

print("Gradient Boosting")
print("------------------")
print("MAE:", round(gradient_mae, 4))
print("RMSE:", round(gradient_rmse, 4))
print("R²:", round(gradient_r2, 4))
```

```python
# MODEL COMPARISON

model_results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest",
        "Gradient Boosting"
    ],
    "MAE": [
        linear_mae,
        rf_mae,
        gradient_mae
    ],
    "RMSE": [
        linear_rmse,
        rf_rmse,
        gradient_rmse
    ],
    "R2": [
        linear_r2,
        rf_r2,
        gradient_r2
    ]
})

display(
    model_results.sort_values(
        "RMSE"
    ).reset_index(drop=True)
)
```

```python
# DATA FORECAST

forecast_data = (
    df.groupby(
        ["Country", "Food Category", "Year"],
        as_index=False
    )
    .agg({
        "Total Waste (Tons)": "sum",
        "Economic Loss (Million $)": "sum",
        "Avg Waste per Capita (Kg)": "mean",
        "Population (Million)": "mean",
        "Household Waste (%)": "mean"
    })
)

display(forecast_data.head())

print(
    "Forecasting dataset shape:",
    forecast_data.shape
)
```

```python
from sklearn.linear_model import LinearRegression


def forecast_food_waste(
    country,
    food_category,
    future_years=[2025, 2026, 2027]
):

    # Select country and category
    subset = forecast_data[
        (forecast_data["Country"] == country) &
        (forecast_data["Food Category"] == food_category)
    ].copy()

    # Sort by year
    subset = subset.sort_values("Year")

    # Check that enough historical data exists
    if len(subset) < 3:
        raise ValueError(
            "Not enough historical observations "
            "for this selection."
        )

    # Training variables
    X_hist = subset[["Year"]]
    y_hist = subset["Total Waste (Tons)"]

    # Train simple time trend model
    forecast_model = LinearRegression()

    forecast_model.fit(
        X_hist,
        y_hist
    )

    # Future years
    future_df = pd.DataFrame({
        "Year": future_years
    })

    future_predictions = (
        forecast_model.predict(
            future_df[["Year"]]
        )
    )

    future_df["Forecast Waste (Tons)"] = (
        future_predictions
    )

    # Do not allow negative waste
    future_df["Forecast Waste (Tons)"] = (
        future_df["Forecast Waste (Tons)"]
        .clip(lower=0)
    )

    # Return both historical and forecast data
    historical = subset[
        ["Year", "Total Waste (Tons)"]
    ].copy()

    historical["Type"] = "Historical"

    historical = historical.rename(
        columns={
            "Total Waste (Tons)": "Waste (Tons)"
        }
    )

    future_output = future_df[
        ["Year", "Forecast Waste (Tons)"]
    ].copy()

    future_output["Type"] = "Forecast"

    future_output = future_output.rename(
        columns={
            "Forecast Waste (Tons)": "Waste (Tons)"
        }
    )

    combined = pd.concat(
        [historical, future_output],
        ignore_index=True
    )

    return combined, forecast_model
```

```python
def plot_forecast(
    country,
    food_category
):

    forecast_result, model = (
        forecast_food_waste(
            country,
            food_category
        )
    )

    historical = forecast_result[
        forecast_result["Type"] == "Historical"
    ]

    future = forecast_result[
        forecast_result["Type"] == "Forecast"
    ]

    plt.figure(figsize=(11, 6))

    plt.plot(
        historical["Year"],
        historical["Waste (Tons)"],
        marker="o",
        label="Historical"
    )

    plt.plot(
        future["Year"],
        future["Waste (Tons)"],
        marker="o",
        linestyle="--",
        label="Forecast"
    )

    plt.axvline(
        x=2024.5,
        linestyle=":"
    )

    plt.text(
        2024.55,
        plt.ylim()[1] * 0.95,
        "Forecast begins",
        rotation=90,
        verticalalignment="top"
    )

    plt.title(
        f"Food Waste Forecast: "
        f"{country} — {food_category}"
    )

    plt.xlabel("Year")
    plt.ylabel("Food Waste (Tons)")

    plt.legend()
    plt.tight_layout()

    plt.show()

    return forecast_result
```

```python
def what_if_simulation(
    waste_tons,
    economic_loss,
    reduction_percentage
):

    reduction_percentage = float(
        reduction_percentage
    )

    waste_reduction = (
        waste_tons *
        reduction_percentage / 100
    )

    projected_waste = (
        waste_tons -
        waste_reduction
    )

    estimated_loss_avoided = (
        economic_loss *
        reduction_percentage / 100
    )

    projected_loss = (
        economic_loss -
        estimated_loss_avoided
    )

    return {
        "Current Waste (Tons)": waste_tons,
        "Waste Potentially Avoided (Tons)": waste_reduction,
        "Projected Waste (Tons)": projected_waste,
        "Current Economic Loss ($M)": economic_loss,
        "Estimated Loss Avoided ($M)": estimated_loss_avoided,
        "Projected Economic Loss ($M)": projected_loss
    }
```

```python
def generate_recommendations(
    waste_per_capita,
    household_waste,
    reduction_percentage
):

    recommendations = []

    if waste_per_capita >= 110:
        recommendations.append(
            "Prioritize actions that reduce "
            "per-person food waste."
        )
    elif waste_per_capita >= 100:
        recommendations.append(
            "Promote better food planning, "
            "storage and portion management."
        )
    else:
        recommendations.append(
            "Continue monitoring per-capita "
            "food waste."
        )

    if household_waste >= 50:
        recommendations.append(
            "Focus on household food planning, "
            "storage, portion control and redistribution."
        )
    else:
        recommendations.append(
            "Continue monitoring household waste "
            "and prevention efforts."
        )

    if reduction_percentage >= 20:
        recommendations.append(
            "The selected reduction scenario represents "
            "a substantial intervention target."
        )
    elif reduction_percentage >= 10:
        recommendations.append(
            "The selected scenario provides a useful "
            "moderate reduction target for comparison."
        )
    else:
        recommendations.append(
            "Test additional reduction scenarios "
            "to compare possible outcomes."
        )

    return recommendations
```

```python
def analyze_scenario(
    country,
    food_category,
    year,
    reduction_percentage
):

    filtered = df[
        (df["Country"] == country) &
        (df["Food Category"] == food_category) &
        (df["Year"] == year)
    ]

    if len(filtered) == 0:
        return "No data found for this selection."

    waste = filtered[
        "Total Waste (Tons)"
    ].sum()

    economic_loss = filtered[
        "Economic Loss (Million $)"
    ].sum()

    waste_per_capita = filtered[
        "Avg Waste per Capita (Kg)"
    ].mean()

    population = filtered[
        "Population (Million)"
    ].mean()

    household_waste = filtered[
        "Household Waste (%)"
    ].mean()

    # What-if scenario
    waste_reduction = (
        waste *
        reduction_percentage /
        100
    )

    projected_waste = (
        waste -
        waste_reduction
    )

    loss_avoided = (
        economic_loss *
        reduction_percentage /
        100
    )

    projected_loss = (
        economic_loss -
        loss_avoided
    )

    recommendations = generate_recommendations(
        waste_per_capita,
        household_waste,
        reduction_percentage
    )

    recommendation_text = "\n".join(
        f"- {r}" for r in recommendations
    )

    result = f"""
# 🌍 FoodWaste AI Report

## Selected Scenario

**Country:** {country}

**Food Category:** {food_category}

**Year:** {year}

---

## 📊 Current Situation

| Indicator | Result |
|---|---:|
| Food Waste | {waste:,.2f} tons |
| Economic Loss | ${economic_loss:,.2f}M |
| Waste per Capita | {waste_per_capita:,.2f} kg |
| Population | {population:,.2f}M |
| Household Waste | {household_waste:,.2f}% |

---

## 🔮 What-If Scenario

**Waste reduction target:** {reduction_percentage}%

### Potential waste avoided

**{waste_reduction:,.2f} tons**

### Projected waste

**{projected_waste:,.2f} tons**

### Estimated economic loss avoided

**${loss_avoided:,.2f} million**

### Projected economic loss

**${projected_loss:,.2f} million**

---

## 💡 AI Recommendations

{recommendation_text}

---

### About the estimate

The scenario uses the selected historical record and
applies the chosen reduction percentage to illustrate
a possible intervention outcome.
"""

    return result
```

```python
def forecast_for_app(
    country,
    food_category
):

    result, model = forecast_food_waste(
        country,
        food_category,
        [2025, 2026, 2027]
    )

    future = result[
        result["Type"] == "Forecast"
    ].copy()

    future["Waste (Tons)"] = (
        future["Waste (Tons)"]
        .round(2)
    )

    forecast_table = future[
        ["Year", "Waste (Tons)"]
    ]

    markdown_table = (
        forecast_table
        .to_markdown(
            index=False
        )
    )

    output = f"""
# 🔮 Food Waste Forecast

**Country:** {country}

**Food Category:** {food_category}

The model uses historical observations from
**2018–2024** to estimate food waste for
**2025–2027**.

## Forecast

{markdown_table}

### Interpretation

These values are model-based estimates from the
historical pattern in the dataset. They are not
observed future values.
"""

    # Create chart
    fig, ax = plt.subplots(
        figsize=(10, 5)
    )

    historical = result[
        result["Type"] == "Historical"
    ]

    future_data = result[
        result["Type"] == "Forecast"
    ]

    ax.plot(
        historical["Year"],
        historical["Waste (Tons)"],
        marker="o",
        label="Historical"
    )

    ax.plot(
        future_data["Year"],
        future_data["Waste (Tons)"],
        marker="o",
        linestyle="--",
        label="Forecast"
    )

    ax.axvline(
        2024.5,
        linestyle=":"
    )

    ax.set_title(
        f"Food Waste Forecast — "
        f"{country}, {food_category}"
    )

    ax.set_xlabel("Year")
    ax.set_ylabel("Food Waste (Tons)")

    ax.legend()

    fig.tight_layout()

    return output, fig
```
